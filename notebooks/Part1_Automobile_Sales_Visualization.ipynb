# IBM DV0101EN – Automobile Sales Analysis
## Visualization notebook
This notebook uses the IBM Skills Network historical automobile sales dataset. The dataset is synthetic and intended for educational use.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

DATA_URL = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv'
df = pd.read_csv(DATA_URL)
df['Date'] = pd.to_datetime(df['Date'])
display(df.head())
print('Shape:', df.shape)
display(df.isna().sum())
```

## 1. Annual automobile sales

```python
yearly = df.groupby('Year', as_index=False)['Automobile_Sales'].mean()
plt.figure(figsize=(11,5))
sns.lineplot(data=yearly, x='Year', y='Automobile_Sales', marker='o')
plt.title('Average Automobile Sales by Year'); plt.grid(True); plt.tight_layout(); plt.show()
```

## 2. Vehicle-type sales during recessions

```python
rec = df[df['Recession'] == 1]
trend = rec.groupby(['Year','Vehicle_Type'], as_index=False)['Automobile_Sales'].mean()
plt.figure(figsize=(12,6))
sns.lineplot(data=trend, x='Year', y='Automobile_Sales', hue='Vehicle_Type', marker='o')
plt.title('Vehicle-Type Sales During Recession'); plt.tight_layout(); plt.show()
```

## 3. Recession vs non-recession comparison

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df, x='Vehicle_Type', y='Automobile_Sales', hue='Recession', errorbar=None)
plt.title('Vehicle Sales by Economic Condition'); plt.tight_layout(); plt.show()
```

## 4. GDP by economic condition

```python
gdp = df.groupby(['Year','Recession'], as_index=False)['GDP'].mean()
sns.lineplot(data=gdp, x='Year', y='GDP', hue='Recession', marker='o')
plt.title('GDP by Year and Recession Status'); plt.tight_layout(); plt.show()
```

## 5. Seasonality, price and advertising

```python
monthly = df.groupby('Month', as_index=False).agg(sales=('Automobile_Sales','mean'), seasonality=('Seasonality_Weight','mean'))
plt.figure(figsize=(10,5)); plt.scatter(monthly['Month'], monthly['sales'], s=monthly['seasonality']*500, alpha=.6)
plt.title('Monthly Sales and Seasonality'); plt.xlabel('Month'); plt.ylabel('Average Sales'); plt.tight_layout(); plt.show()

plt.figure(figsize=(8,5)); plt.scatter(rec['Price'], rec['Automobile_Sales'], alpha=.5)
plt.title('Vehicle Price vs Sales During Recession'); plt.tight_layout(); plt.show()

ad = df.groupby('Recession', as_index=False)['Advertising_Expenditure'].sum()
display(ad)
```

## 6. Unemployment and vehicle sales

```python
unemp = rec.groupby(['unemployment_rate','Vehicle_Type'], as_index=False)['Automobile_Sales'].mean()
plt.figure(figsize=(11,5)); sns.lineplot(data=unemp, x='unemployment_rate', y='Automobile_Sales', hue='Vehicle_Type')
plt.title('Unemployment Rate and Vehicle Sales'); plt.tight_layout(); plt.show()
```
